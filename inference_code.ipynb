# Question 7 - Ames Housing Dataset

7. In `./data/ames_prices.csv`, create an indicator that equals 1 when `Central.Air == 'Y'` and 0 otherwise. 
  

```python
# Problem 7(a)

air = (df["Central.Air"] == "Y").astype(int)
p_hat = air.mean()
print(f"Estimated proportion with central air: {100*p_hat:.2f}%")

# Interpretation (a):
# We estimate that approximately 93.12% of Ames homes have central air.
```

```python
# Problem 7(b)

seed = 100
B = 5000

rng = np.random.default_rng(seed=seed)

boot_props = []

for b in range(B):

    sample = rng.choice(
        air,
        size=len(air),
        replace=True
    )

    boot_props.append(sample.mean())

boot_props = np.array(boot_props)

print(f"Random seed used: {seed}")
print(f"Number of bootstrap samples: {B}")

# Interpretation (b):
# We generated 5,000 bootstrap samples using random seed 100.
# Each bootstrap sample was created by sampling homes with replacement from the original Ames dataset.
```

```python
# Problem 7(c)

SE_boot = boot_props.std(ddof=1)

print(f"Bootstrap standard error: {100*SE_boot:.2f}%")

# Interpretation (c):
# The bootstrap standard error is 0.46%. This means the estimated percentage of Ames homes
# with central air would typically vary by about 0.46 percentage points from sample to sample.
```

```python
# Problem 7(d)

CI90 = (
    p_hat - 1.645*SE_boot,
    p_hat + 1.645*SE_boot
)

CI95 = (
    p_hat - 1.96*SE_boot,
    p_hat + 1.96*SE_boot
)

CI99 = (
    p_hat - 2.576*SE_boot,
    p_hat + 2.576*SE_boot
)

print(
    f"90% CI: ({100*CI90[0]:.2f}%, "
    f"{100*CI90[1]:.2f}%)"
)

print(
    f"95% CI: ({100*CI95[0]:.2f}%, "
    f"{100*CI95[1]:.2f}%)"
)

print(
    f"99% CI: ({100*CI99[0]:.2f}%, "
    f"{100*CI99[1]:.2f}%)"
)

# Interpretation (d):
# We are 90% confident that between 92.55% and 94.07% of Ames homes have central air.
# We are 95% confident that between 92.41% and 94.21% of Ames homes have central air.
# We are 99% confident that between 92.12% and 94.50% of Ames homes have central air.
```

```python
# Problem 7(e)

print("90% in 90% CI:", CI90[0] <= 0.90 <= CI90[1])
print("90% in 95% CI:", CI95[0] <= 0.90 <= CI95[1])
print("90% in 99% CI:", CI99[0] <= 0.90 <= CI99[1])

# Interpretation (e):
# We reject H0: p = 90% at the 10%, 5%, and 1% significance levels. This is because 90% falls outside all three confidence intervals.
# We estimate that approximately 93.31% of Ames homes have central air.
# The data provide strong evidence that the true percentage of Ames homes with central air is greater than 90%.
```

# Question 8 - Pearson correlation 

```python
# Problem 8(a)

df = pd.read_csv("./data/ames_prices.csv")

x = np.arcsinh(df["area"].values)
y = np.arcsinh(df["price"].values)

corr_hat = np.corrcoef(x,y)[0,1]

print(f"Estimated correlation: {corr_hat:.3f}")

seed = 100
B = 5000

rng = np.random.default_rng(seed)

boot_corrs = []

for b in range(B):

    idx = rng.choice(
        np.arange(len(df)),
        size=len(df),
        replace=True
    )

    corr_b = np.corrcoef(
        x[idx],
        y[idx]
    )[0,1]

    boot_corrs.append(corr_b)

boot_corrs = np.array(boot_corrs)

print("Bootstrap estimates:", len(boot_corrs))
```

```python
# Problem 8(b)
print(len(boot_corrs))
SE_boot = boot_corrs.std(ddof=1)

print(f"Bootstrap standard error: {SE_boot:.4f}")
```

```python
# Problem 8(c)

corr_hat = 0.723
SE_boot = 0.0102

CI90 = (
    corr_hat - 1.645*SE_boot,
    corr_hat + 1.645*SE_boot
)

CI95 = (
    corr_hat - 1.96*SE_boot,
    corr_hat + 1.96*SE_boot
)

CI99 = (
    corr_hat - 2.576*SE_boot,
    corr_hat + 2.576*SE_boot
)

print(f"90% CI: ({CI90[0]:.3f}, {CI90[1]:.3f})")
print(f"95% CI: ({CI95[0]:.3f}, {CI95[1]:.3f})")
print(f"99% CI: ({CI99[0]:.3f}, {CI99[1]:.3f})")

# Interpretation (c):
# We are 90% confident that the population correlation lies between 0.706 and 0.740. 
# We are 95% confident that the population correlation lies between 0.703 and 0.743.
# We are 99% confident that the population correlation lies between 0.697 and 0.749.
```

```python
# Problem 8(d)

print("0 in 90% CI:",
      CI90[0] <= 0 <= CI90[1])

print("0 in 95% CI:",
      CI95[0] <= 0 <= CI95[1])

print("0 in 99% CI:",
      CI99[0] <= 0 <= CI99[1])

# Interpretation (d): We reject H0: rho = 0 at the 10%, 5%, and 1% significance levels.
# This is because 0 does not fall within any of the confidence intervals.
# We estimate a correlation of 0.723 between transformed home area and transformed home price.
# Homes with larger living areas tend to be associated with higher sale prices.
# This result describes an association only and should not be interpreted as a causal relationship.
```

# Interpretation 8(e):
Area and price must remain paired because each row represents a single home. Resampling rows preserves the observed relationship between a home's area and its sale price. If area and price were resampled separately, we would create combinations of areas and prices that never occurred in the original dataset. This would destroy the relationship we are trying to measure and would produce an incorrect estimate of the correlation.

# Question 9 - Overall Survival (Months) 

```python
# Problem 9(a)

import pandas as pd
import numpy as np

df = pd.read_csv("./data/metabric.csv")

survival = df["Overall Survival (Months)"]

mean_survival = survival.mean()

print(f"Estimated mean survival/follow-up time: {mean_survival:.2f} months")

seed = 100
B = 5000

rng = np.random.default_rng(seed)

boot_means = []

for b in range(B):

    sample = rng.choice(
        survival,
        size=len(survival),
        replace=True
    )

    boot_means.append(sample.mean())

boot_means = np.array(boot_means)

print(f"Random seed used: {seed}")
print(f"Bootstrap replications: {B}")

# Interpretation (a):
# We estimate that the average recorded survival or follow-up time is 129.09 months.
# We generated 5,000 bootstrap samples using random seed 100.
```

```python
# Problem 9(b)

SE_boot = boot_means.std(ddof=1)

print(f"Bootstrap standard error: {SE_boot:.2f} months")
```

```python
# Problem 9(c)

mean_survival = 129.09
SE_boot = 2.12

CI90 = (
    mean_survival - 1.645*SE_boot,
    mean_survival + 1.645*SE_boot
)

CI95 = (
    mean_survival - 1.96*SE_boot,
    mean_survival + 1.96*SE_boot
)

CI99 = (
    mean_survival - 2.576*SE_boot,
    mean_survival + 2.576*SE_boot
)

print(f"90% CI: ({CI90[0]:.2f}, {CI90[1]:.2f}) months")
print(f"95% CI: ({CI95[0]:.2f}, {CI95[1]:.2f}) months")
print(f"99% CI: ({CI99[0]:.2f}, {CI99[1]:.2f}) months")

# Interpretation (c):
# We are 90% confident that the average recorded survival or follow-up time is between 125.60 and 132.58 months. We are 95% confident that the average recorded survival or follow-up time is between 124.93 and 133.25 months.
# We are 99% confident that the average recorded survival or follow-up time is between 123.63 and 134.55 months.
```

```python
# Problem 9(d)

print("120 in 90% CI:",
      CI90[0] <= 120 <= CI90[1])

print("120 in 95% CI:",
      CI95[0] <= 120 <= CI95[1])

print("120 in 99% CI:",
      CI99[0] <= 120 <= CI99[1])

# Interpretation (d):
# We reject H0: mu = 120 months at the 10%, 5%, and 1% significance levels.
# This is because 120 months does not fall within any of the confidence intervals. 
# The data provide strong evidence that the average recorded survival or follow-up time exceeds 120 months.
```

```python
# Problem 9(e)

print(f"Mean survival time: {129.09:.2f} months")
print(f"Mean survival time: {129.09/12:.2f} years")
```

# Interpretation 9(e):
We estimate that the average recorded survival or follow-up time is 129.09 months, which is approximately 10.76 years. The confidence intervals suggest that the true average recorded survival or follow-up time is between about 124 and 135 months, or roughly 10.4 to 11.2 years. These results should not be interpreted as the average lifetime of patients. Some patients were still alive when the data were collected, so their eventual death times are unknown. Because these observations are right-censored, the analysis estimates average observed survival or follow-up time rather than unrestricted lifetime.

# Problem 10 - Sleep Hours

```python
# Problem 10(a)

df = pd.read_csv('./data/nhanes_data_17_18.csv')

adults = df[
    (df['AgeInYearsAtScreening'] >= 18)
    & df['SleepHoursWeekdaysOrWorkdays'].notna()
]

x = adults['SleepHoursWeekdaysOrWorkdays'].to_numpy()

n = len(x)

mu_hat = x.mean()

seed = 10
R = 2000

rng = np.random.default_rng(seed)

boot = np.array([
    rng.choice(
        x,
        n,
        replace=True
    ).mean()
    for _ in range(R)
])

print(f'Seed: {seed}')
print(f'Replications: {R}')
print(f'Sample size: {n}')
print(f'Estimated mean weekday sleep: {mu_hat:.3f} hours')

# Interpretation (a):
# We estimate that adults in this sample sleep
# an average of 7.621 hours on weekdays or workdays.
# We generated 2,000 bootstrap samples using random seed 10.
```

```python
# Problem 10(b)

se = boot.std(ddof=1)

print(
    f'Bootstrap standard error: '
    f'{se:.4f} hours'
)

# Interpretation (b):
# The bootstrap standard error is 0.0225 hours.
# This means the estimated average sleep duration would typically vary by about 0.0225 hours from sample to sample.
```

```python
# Problem 10(c)

mu_hat = 7.621
SE_boot = 0.0225

CI90 = (
    mu_hat - 1.645*SE_boot,
    mu_hat + 1.645*SE_boot
)

CI95 = (
    mu_hat - 1.96*SE_boot,
    mu_hat + 1.96*SE_boot
)

CI99 = (
    mu_hat - 2.576*SE_boot,
    mu_hat + 2.576*SE_boot
)

print(f"90% CI: ({CI90[0]:.3f}, {CI90[1]:.3f}) hours")
print(f"95% CI: ({CI95[0]:.3f}, {CI95[1]:.3f}) hours")
print(f"99% CI: ({CI99[0]:.3f}, {CI99[1]:.3f}) hours")

# Interpretation (c):
# We are 90% confident that the average weekday sleep duration is between 7.584 and 7.658 hours.
# We are 95% confident that the average weekday sleep duration is between 7.577 and 7.665 hours.
# We are 99% confident that the average weekday sleep duration is between 7.563 and 7.679 hours.
```

```python
# Problem 10(d)

print("8 in 90% CI:",
      CI90[0] <= 8 <= CI90[1])

print("8 in 95% CI:",
      CI95[0] <= 8 <= CI95[1])

print("8 in 99% CI:",
      CI99[0] <= 8 <= CI99[1])

# Interpretation (d):
# We reject H0: mu = 8 hours at the 10%, 5%, and 1% significance levels.
# This is because 8 hours does not fall within any of the confidence intervals.
# The data provide strong evidence that the average weekday sleep duration differs from 8 hours.
```

# Interpretation (e)

We estimate that adults in this sample sleep an average of **7.62 hours** on weekdays or workdays.

We reject \(H_0:\mu=8\) hours because **8 hours falls outside all of the confidence intervals**.

The data provide strong evidence that average weekday sleep is **less than 8 hours**. However, the estimated difference is only about **0.38 hours (23 minutes)**, so while the result is statistically significant, its practical importance may be limited.

These results are based on an unweighted sample and should not be interpreted as nationally representative.
